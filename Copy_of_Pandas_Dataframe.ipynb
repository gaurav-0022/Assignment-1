## Q1. Creating a DataFrame

```python
import pandas as pd

data = {
    "Product": ["A", "B", "C", "D", "E"],
    "Region": ["East", "West", "North", "South", "East"],
    "Sales": [1000, 1500, 2000, 1200, 1800],
    "Quantity": [10, 15, 20, 12, 18]
}

df1 = pd.DataFrame(data)
df1
```

##Load Dataset (for Q2–Q10)

```python
df = pd.read_csv("Retail_dataset.csv")
df.head()
```

##Q2: Data Inspection

```python
# First 5 rows
df.head()

# Last 5 rows
df.tail()

# Dataset info
df.info()

# Shape
df.shape
```

## Q3. Column Selection

```python
df[["Product", "Region", "Sales"]]
```

## Q4. Indexing using .loc

```python
# Row with index 10
df.loc[10]

# Sales value at index 20
df.loc[20, "Sales"]
```

## Q5. Indexing using .iloc

```python
# First 6 rows and first 4 columns
df.iloc[:6, :4]

# Sales value from 3rd row
df.iloc[2]["Sales"]
```

## Q6. Row Slicing

```python
# First 10 rows
df.head(10)

# Rows 15 to 25
df.iloc[15:26]
```

## Q7. Filtering Data

```python
# Region = East
df[df["Region"] == "East"]

# Sales > 1500
df[df["Sales"] > 1500]
```

## Q8. Multiple Condition Filtering

```python
df[(df["Region"] == "West") & (df["Sales"] > 1200)]
```

## Q9. Descriptive Statistics

```python
stats = df.describe()
stats

# Extract values
mean_sales = stats.loc["mean", "Sales"]
max_quantity = stats.loc["max", "Quantity"]
min_sales = stats.loc["min", "Sales"]

print("Mean Sales:", mean_sales)
print("Max Quantity:", max_quantity)
print("Min Sales:", min_sales)
```

## Q10. Aggregation Analysis

```python
total_sales = df["Sales"].sum()
avg_sales = df["Sales"].mean()
max_sales = df["Sales"].max()
min_sales = df["Sales"].min()

print("Total Sales:", total_sales)
print("Average Sales:", avg_sales)
print("Maximum Sales:", max_sales)
print("Minimum Sales:", min_sales)
```
